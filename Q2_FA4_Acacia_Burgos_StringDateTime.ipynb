## 📝 Activity 1: Student Information Formatter

**Task:** Create a program that asks for student details and displays formatted output according to PSHS standards.

**Required Fields:**
1. First Name
2. Last Name
3. Section
4. Favorite Subject



```python
# Activity 1 Implementation Code
# Test with messy inputs like '  juan ', ' dela cruz ', 'narra ',' acacia ', 'mahogany ', '  molave ' 'computer science'

first_name = input(str("Enter your first name: "))
last_name = input(str("Enter your last name: "))
section = input(str("Enter your section: "))
fav_subject = input(str("Enter your favorite subject: "))

# Cleaning and formatting logic
#### Add your Code here ####
full_name = first_name.strip().title()+ " " + last_name.strip().title()
full_name_upper=first_name.strip().upper()+ " "+ last_name.strip().upper()
formatted_section=section.strip().upper()
formatted_subject=section.strip().upper()
char_count=len(full_name)

print("--- Formatted Output ---")
print(f"Name: {full_name}")
print(f"Name Code: {full_name_upper}")
print(f"Section: {formatted_section}")
print(f"Favorite Subject: {formatted_subject}")
print(f"Character Count: {char_count}")

```

## 📅 Activity 2: Deadline & Date Formatter

**Task:** Prompt the user for an event title, year, month, and day, then output a clean schedule card with the weekday!



```python
# Activity 2: Deadline & Date Formatter
# Input Data from user
from datetime import datetime

event_title = str(input("Enter event title: "))
year = int(input("Enter event year: "))
month = int(input("Enter event month: "))
day = int(input("Enter event day: "))

# Create datetime object
event_date=datetime(year, month, day)


print("--- Formatted Output ---")
print(f"Event: {event_title.strip().title()}")
print(f"Date: {event_date.strftime("%B %d, %Y")}")
print(f"Year: {year}")
print(f"Month: {event_date.strftime("%B")}")
print(f"Day: {day}")
print(f"Weekday: {event_date.strftime("%A")}")

```
