```python
scores = [85, 90, 78, 92, 88]

while True:
    print("\n1. Display")
    print("2. Search")
    print("3. Average")
    print("4. Update")
    print("5. Exit")

    choice = int(input("Choice(1, 2, 3, 4, 5): "))

    if choice == 1:
        print(scores)

    elif choice == 2:
        x = int(input("Score to search: "))
        if x in scores:
            print("Found")
        else:
            print("Not Found")

    elif choice == 3:
        print("Average =", sum(scores) / len(scores))

    elif choice == 4:
        i = int(input("Index: "))
        n = int(input("New score: "))
        scores[i] = n
        print("Updated:", scores)

    elif choice == 5:
        break
```
