```python
#Activity 1
number = ["Zero", "One", "Two", "Three", "Four"]

print("--- Direct Index Access ---")
print("Element at index 0:",  number[0])
print("Element at index 1:",  number[1])
print("Element at index 2:",  number[2])
print("Element at index 3:", number[3])
print("Element at index 4:", number[4])

print("\n--- Modifying an Element via Index ---")
print("Original index 2 item:", number[2])

number[2] = "Three"

print("Updated index 2 item:", number[2])
print("Updated number List:", number)
```

```python
#Activity 2
items = ["Notebook", "Pen", "Calculator", "Laptop", "Book"]

print("--- Direct Index Access ---")
print("Element at index 0:", items[0])
print("Element at index 1:", items[1])
print("Element at index 2:", items[2])
print("Element at index 3:", items[3])
print("Element at index 4:", items[4])

print("Original Array/List:", items)

index = int(input("Index to Update: "))
new_value = input("New Value: ")

start = int(input("Slice Start: "))
end = int(input("Slice End: "))

slice_list = items[start:end]

search = input("Search Item: ")

print("Sliced Array/List:", slice_list)

if search in slice_list:
    print(search, "is found in the sliced list.")
else:
    print(search, "is not found in the sliced list.")

items[index] = new_value

print("Updated Array/List:", items)
```

```python
#Activity 3
scores = [85, 90, 78, 92, 88]

while True:
    print("\n1. Display")
    print("2. Search")
    print("3. Average")
    print("4. Update")
    print("5. Exit")

    choice = int(input("Choice(1, 2, 3, 4, 5): "))

    if choice == 1:
        print(scores)

    elif choice == 2:
        x = int(input("Score to search: "))
        if x in scores:
            print("Found")
        else:
            print("Not Found")

    elif choice == 3:
        print("Average =", sum(scores) / len(scores))

    elif choice == 4:
        i = int(input("Index: "))
        n = int(input("New score: "))
        scores[i] = n
        print("Updated:", scores)

    elif choice == 5:
        break
```
