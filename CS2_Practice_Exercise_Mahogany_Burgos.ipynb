```python
#Problem 1
def update_savings(total, deposit):
    return total + deposit

def check_goal(total, target):
    if total >= target:
        return "Goal reached!"
    return "Keep saving!"

target = float(input("Enter savings target: ₱"))
total = 0

deposit = float(input("Enter deposit (0 to finish): ₱"))

while deposit != 0:
    total = update_savings(total, deposit)
    print("Updated savings: ₱", format(total, ".2f"))
    deposit = float(input("Enter deposit (0 to finish): ₱"))

print("Total savings: ₱", format(total, ".2f"))

if total < target:
    print("Remaining amount needed: ₱", format(target - total, ".2f"))

print(check_goal(total, target))
```

```python
#Problem 2
def calculate_energy(w, h):
    return w * h / 1000

def classify_usage(e):
    if e < 0.5:
        return "Low"
    elif e <= 2:
        return "Moderate"
    else:
        return "High"

n = int(input("How many devices? "))

total = 0
highest = 0
device = ""

for i in range(n):
    name = input("Name: ")
    w = float(input("Watts: "))
    h = float(input("Hours: "))

    e = calculate_energy(w, h)

    print("Energy:", round(e, 2), "kWh")
    print(classify_usage(e))

    total += e

    if e > highest:
        highest = e
        device = name

print("Total:", round(total, 2), "kWh")
print("Highest:", device)
```

```python
#Problem 3
def calculate_sale(qty, price):
    return qty * price

def calculate_discount(total):
    if total >= 500:
        return total * 0.05
    return 0

price = float(input("Price per container: ₱"))

sales = 0
discounts = 0

qty = int(input("Quantity sold (0 to finish): "))

while qty != 0:
    sale = calculate_sale(qty, price)
    discount = calculate_discount(sale)

    print("Sale: ₱", sale)
    print("Discount: ₱", discount)
    print("Amount Due: ₱", sale - discount)

    sales += sale
    discounts += discount

    qty = int(input("Quantity sold (0 to finish): "))

print("Total Sales: ₱", sales)
print("Total Discounts: ₱", discounts)
print("Net Sales: ₱", sales - discounts)
```

```python
#Problem 4
def calculate_reading_speed(pages, minutes):
    if minutes == 0:
        return 0
    return pages / minutes

def classify_speed(speed):
    if speed < 1:
        return "Developing"
    elif speed <= 2:
        return "Steady"
    return "Fast"

n = int(input("How many sessions? "))

total_pages = 0
total_minutes = 0

for i in range(n):
    pages = int(input("Pages read: "))
    minutes = int(input("Minutes: "))

    speed = calculate_reading_speed(pages, minutes)

    print("Speed:", round(speed, 2))
    print(classify_speed(speed))

    total_pages += pages
    total_minutes += minutes

overall = calculate_reading_speed(total_pages, total_minutes)

print("Total pages:", total_pages)
print("Total minutes:", total_minutes)
print("Overall speed:", round(overall, 2))
```

```python
#Problem 5
def calculate_attendance_rate(a, e):
    if e == 0:
        return 0
    return (a / e) * 100

def classify_attendance(r):
    if r < 60:
        return "Low"
    elif r < 90:
        return "Moderate"
    return "High"

n = int(input("How many sessions? "))

total_a = 0
total_e = 0
high = 0

for i in range(n):
    e = int(input("Expected attendance: "))
    a = int(input("Actual attendees: "))

    rate = calculate_attendance_rate(a, e)

    print("Rate:", round(rate, 2), "%")
    print(classify_attendance(rate))

    if rate >= 90:
        high += 1

    total_a += a
    total_e += e

overall = calculate_attendance_rate(total_a, total_e)

print("Total attendees:", total_a)
print("Total expected:", total_e)
print("Overall rate:", round(overall, 2), "%")
print("High sessions:", high)
```
